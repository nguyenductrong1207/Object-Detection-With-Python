# Start

⚡ Login with your API key, load your YOLO 🚀 model, and start training in 3 lines of code!

```python
import torch
import psutil
from ultralytics import YOLO, checks, hub
checks()

hub.login('95599867f2d95b986d06ed6243f0c1276ca9f0fbb1')

# Callback log VRAM, CPU, RAM sau mỗi epoch
def log_system_usage(trainer):
    # GPU (VRAM)
    used = torch.cuda.memory_allocated() / 1024 ** 2  # MB
    reserved = torch.cuda.memory_reserved() / 1024 ** 2  # MB

    # CPU & RAM
    cpu_usage = psutil.cpu_percent(interval=None)  # %
    ram_usage = psutil.virtual_memory().used / 1024 ** 2  # MB

    log_line = (f"Epoch {trainer.epoch + 1}/{trainer.epochs} | "
                f"VRAM used: {used:.2f} MB, reserved: {reserved:.2f} MB | "
                f"CPU: {cpu_usage:.1f}% | RAM: {ram_usage:.2f} MB\n")

    # Ghi log ra file
    with open("system_usage_log.txt", "a") as f:
        f.write(log_line)

    # In ra console
    print("[Monitor]", log_line.strip())

# Load model YOLO11
model = YOLO('https://hub.ultralytics.com/models/mYYENpkQFBGAkuTV6Ox8')

# Train với callback
results = model.train(
    data="data.yaml",   # đường dẫn file dataset
    imgsz=1024,         # resize ảnh về 640x640
    epochs=100,         # số epoch
    batch=4,            # batch size vừa với 6GB VRAM (4–8), imgsz=1024 => batch 2-4
    workers=6,          # số CPU worker load data
    device=0,           # train trên GPU (train trên CPU => "cpu")
    callbacks={"on_train_epoch_end": log_system_usage}  # log sau mỗi epoch
)

```
